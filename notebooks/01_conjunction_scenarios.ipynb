# Conjunction scenarios

The three `ConjunctionScenario` presets, plotted with the functions in `spacecraft_belief_reachability.plotting`.

- `crossing`: circular debris on a nearby plane; slow (~0.7 km/s), with repeat passes every half orbit.
- `high_angle`: circular debris at 97.6° inclination; fast (~6.9 km/s), with repeat passes.
- `eccentric_debris`: 450 × 850 km debris; a single isolated encounter.

Building each scenario takes about 6 s (two propagations with `leo_default()` forces).

```python
%matplotlib inline

import matplotlib.pyplot as plt

from spacecraft_belief_reachability.plotting import (
    plot_conjunction,
    plot_orbits_3d,
    save_figure,
)
from spacecraft_belief_reachability.simulation import ConjunctionScenario

THEME = "light"  # "dark" for slides
SAVE_DIR = None  # e.g. "../figures" to write PDF/SVG/PNG copies
```

```python
scenarios = {
    "crossing": ConjunctionScenario.crossing(),
    "high_angle": ConjunctionScenario.high_angle(),
    "eccentric_debris": ConjunctionScenario.eccentric_debris(),
}
tcas = {key: s.find_tca() for key, s in scenarios.items()}
```

## Summary

TCA, miss distance, relative speed, and the next-closest pass in the propagated span.

```python
print(
    f"{'Scenario':18s} {'TCA (h)':>8s} {'Miss (m)':>9s} {'Speed (km/s)':>13s} {'Next pass (km)':>15s}"
)
for key, s in scenarios.items():
    tca = tcas[key]
    others = [c.miss_m for c in s.close_approaches() if abs(c.t_s - tca.t_s) > 1.0]
    print(
        f"{s.name:18s} {tca.t_s / 3600:8.3f} {tca.miss_m:9.1f} "
        f"{tca.relative_speed_m_s / 1e3:13.2f} {min(others) / 1e3:15.1f}"
    )
```

## Crossing

```python
s, tca = scenarios["crossing"], tcas["crossing"]
fig = plot_conjunction(s, tca=tca, theme=THEME)
if SAVE_DIR:
    save_figure(fig, f"{SAVE_DIR}/crossing_relative")
plt.show()
```

```python
fig = plot_orbits_3d(s, tca=tca, theme=THEME)
if SAVE_DIR:
    save_figure(fig, f"{SAVE_DIR}/crossing_orbits")
plt.show()
```
